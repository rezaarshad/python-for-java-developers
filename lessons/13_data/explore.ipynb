# Exploring the notes data

Run cells with **Shift+Enter**. DataFrames and charts render inline. Try changing things!

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/notes.csv", parse_dates=["created_at"])
df.head()
```

```python
df.info()
```

```python
df.describe()
```

## Clean & enrich

```python
df["tags"] = df["tags"].str.split(";")
df["month"] = df["created_at"].dt.to_period("M")
df["reading_min"] = (df["word_count"] / 200).round(1)
df.sample(5, random_state=0)
```

## Questions to answer

Each cell is one question. Try writing your own below.

```python
# Which notes are longest?
df.nlargest(5, "word_count")[["title", "word_count", "reading_min"]]
```

```python
# How many notes per tag?
df.explode("tags")["tags"].value_counts()
```

```python
# Notes per month, as a chart
df.groupby("month").size().plot(kind="bar", title="Notes per month");
```

```python
# Word count distribution
df["word_count"].plot(kind="hist", bins=15, title="Words per note");
```
